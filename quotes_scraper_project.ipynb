# Quotes scraper project
## Goal of the project
Build a complete data collection pipeline by combining **web scraping**, **S3** object storage (**Minio**), and a NoSQL database (**MongoDB**).

Target site: https://quotes.toscrape.com

## Technical Environment

- Docker and Docker Compose
- Python 3.10+
- Internet Connection

## Target architecture
```
┌─────────────────────┐
│ quotes.toscrape.com │
│                     │
│ • Quotes            │
│ • Authors           │
│ • Tags              │
└──────────┬──────────┘ 
           │ Scraping 
           ▼
┌─────────────────────┐
│       Scraper       │
│       Python        │
└──────────┬──────────┘ 
           │ 
     ┌─────┴─────┐ 
     │           │ 
     ▼           ▼
┌─────────┐ ┌─────────┐
│ MinIO   │ │ MongoDB │
│         │ │         │
│ Exports │ │ Quotes  │
│ Backups │ │ Authors │
│ Images  │ │ Tags    │
└─────────┘ └─────────┘ 
          │ │ 
     └─────┬─────┘ 
           ▼
┌─────────────────────┐
│   Analytics & NLP   │
│     ML Datasets     │
└─────────────────────┘
```

## Tools
### ETL
- Python 3.10+
- Library Beautiful Soup
### Storage
| Component | Role | Data Type |
|-----------|------|-----------------|
| **MinIO** | S3-compatible object storage | Images, raw files, large exports |
| **MongoDB** | NoSQL document database | Structured metadata, queryable data |
